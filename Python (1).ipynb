Question-1. What is the difference between a function and a method in Python?

Answer-1. In Python, both functions and methods are blocks of code that can be called multiple times from different parts of a program. However, there are key differences between them:

Function
1. Definition: A function is a self-contained block of code that performs a specific task.
2. Scope: Functions are defined outside of any class and can be used independently.
3. Call: Functions are called using their name followed by parentheses containing any required arguments.

Method
1. Definition: A method is a function that is defined inside a class and is used to perform operations on instances of that class.
2. Scope: Methods are bound to the class and are used to interact with the instance variables of the class.
3. Call: Methods are called on an instance of the class using dot notation.

Key differences:
- Association: A function is a standalone block of code, while a method is associated with a class and its instances.
- Access to instance variables: Methods have access to the instance variables of the class, while functions do not.
- Call syntax: Functions are called using their name, while methods are called using dot notation on an instance of the class.

Iterators
- Definition: An iterator is an object that keeps track of its position in a sequence and returns the next element each time it is called.
- Examples: Iterators are created using the iter() function or by defining a custom iterator class that implements the iterator protocol (__iter__() and __next__() methods).
- Characteristics: Iterators can be used to iterate over a sequence, and they keep track of their position, allowing them to return the next element on each call.

Example:


my_list = [1, 2, 3]

my_iter = iter(my_list)

print(next(my_iter))  # Output: 1

print(next(my_iter))  # Output: 2

print(next(my_iter))  # Output: 3


Key differences:
- Iterability: Iterables can be iterated over, while iterators are the objects that perform the iteration.
- State: Iterators keep track of their position in the sequence, while iterables do not.
- Usage: Iterables can be used directly in for loops, while iterators are often used when manual control over the iteration process is needed.

Custom Iterator Example:


class MyIterator:
    def __init__(self, data):
        self.data = data
        self.index = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.index < len(self.data):
            result = self.data[self.index]
            self.index += 1
            return result
        raise StopIteration

my_iter = MyIterator([1, 2, 3])
for item in my_iter:
    print(item)


In summary, iterables are objects that can be iterated over, while iterators are objects that perform the iteration and keep track of their position in the sequence. Understanding the difference between iterables and iterators is essential for working with sequences and loops in Python.

Question-6. Explain the concept of generators in Python and how they are defined.

Answer-6. In Python, generators are a type of iterable that can be used to generate a sequence of values on-the-fly, rather than computing them all at once and storing them in memory.

Defining Generators
Generators are defined using functions and the yield keyword. When a generator function is called, it returns a generator iterator.

How Generators Work
1. Execution: When a generator function is called, it executes until it reaches a yield statement.
2. Yielding: The yield statement returns a value and suspends the function's execution.
3. Resuming: When next() is called on the generator iterator, the function resumes execution from where it left off.

Benefits of Generators
1. Memory Efficiency: Generators use less memory because they generate values on-the-fly, rather than storing them all in memory at once.
2. Flexibility: Generators can be used to implement complex iteration logic and can be composed together to create more complex generators.

Generator Expression
Generator expressions are a concise way to define generators using a syntax similar to list comprehensions.

Use Cases for Generators
1. Handling Large Datasets: Generators are useful when working with large datasets that don't fit into memory.
2. Implementing Cooperative Multitasking: Generators can be used to implement cooperative multitasking, where tasks yield control to other tasks voluntarily.
3. Creating Infinite Sequences: Generators can be used to create infinite sequences, such as a sequence of random numbers.

map() Function
- Purpose: Apply a function to each item of an iterable and return a map object.
- Syntax: map(function, iterable)

reduce() Function

- Syntax: reduce(function, iterable)
- Location: reduce() is located in the functools module, so you need to import it: from functools import reduce
- Example: reduce(lambda x, y: x + y, [1, 2, 3, 4]) returns 10

Key differences:
- map(): Transforms each item of an iterable.
- filter(): Selects items from an iterable based on a condition.
- reduce(): Reduces an iterable to a single output value.

When to use each function:
- map(): Use when you need to perform a simple transformation on each item of an iterable.
- filter(): Use when you need to select items from an iterable based on a condition.
- reduce(): Use when you need to reduce an iterable to a single output value, such as calculating a sum or product.

Examples:

numbers = [1, 2, 3, 4, 5]

Question-4. Write a Python function that checks if a given number is prime or not from 1 to 200.

Answer-4. Here's a Python function that checks if a given number is prime or not:


def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True


Here's an explanation of how the code works:

1. The function is_prime takes an integer n as input.
2. If n is less than or equal to 1, it's not prime, so we return False.
3. We iterate from 2 to the square root of n (inclusive) and check if n is divisible by any of these numbers.
4. If n is divisible by any of these numbers, it's not prime, so we return False.
5. If n is not divisible by any of these numbers, it's prime, so we return True.

Example usage:


for i in range(1, 201):

    if is_prime(i):

        print(f"{i} is prime")


This code will print all prime numbers from 1 to 200.

Optimization:

The reason we only need to check up to the square root of n is that a larger factor of n would be a multiple of a smaller factor that has already been checked.

Alternative implementation:

You can also use a more concise approach using a list comprehension:


def is_prime(n):

    return n > 1 and all(n % i != 0 for i in range(2, int(n ** 0.5) + 1))


Both implementations produce the same result, but the first one is often more readable.

Question-5. Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of terms.

Amswer-5. Here's an iterator class in Python that generates the Fibonacci sequence up to a specified number of terms:


class FibonacciIterator:

    def __init__(self, n):
        self.n = n
        self.a = 0
        self.b = 1
        self.current_term = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.current_term >= self.n:
            raise StopIteration
        if self.current_term == 0:
            result = self.a
        else:
            result = self.b
            self.a, self.b = self.b, self.a + self.b
        self.current_term += 1
        return result

# Example usage:

fib_iterator = FibonacciIterator(10)

for num in fib_iterator:

    print(num)


Here's an explanation of how the code works:

1. The FibonacciIterator class takes an integer n as input, which specifies the number of terms in the Fibonacci sequence.
2. The __iter__ method returns the iterator object itself.
3. The __next__ method returns the next term in the Fibonacci sequence. If there are no more terms, it raises a StopIteration exception.
4. In the __next__ method, we calculate the next term in the sequence by adding the previous two terms (self.a and self.b).
5. We use self.current_term to keep track of the current term number.

The example usage will print the first 10 terms of the Fibonacci sequence:

This iterator class is efficient because it only calculates each term once and stores the previous two terms to calculate the next term.

def read_file_line_by_line(filename):

    try:

        with open(filename, 'r') as file:

            for line in file:

                yield line.strip()

    except FileNotFoundError:

        print(f"File '{filename}' not found.")

# Example usage:

for line in read_file_line_by_line('example.txt'):

    print(line)


Here's an explanation of how the code works:

3. The for loop iterates over each line in the file.
4. The yield statement produces each line as a string, stripped of leading/trailing whitespace using strip().
5. If the file doesn't exist, a FileNotFoundError is caught and an error message is printed.

Benefits:

2. Flexibility: You can use this generator function to process files of any size.
